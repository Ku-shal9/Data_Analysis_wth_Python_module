## Imputing values

```python
df.isnull().sum()
```

#### fill missing diameter using absolute magnitude and albedo
#### for this we need the albedo metric and absolute metric to be non-zero
#### we are using the scientific formula, wherever it is possible to fill the metric

```python
mask = (
    df["diameter"].isna()
    & df["absolute_magnitude"].notna()
    & df["albedo"].notna()
    & (df["albedo"] > 0)
)

df.loc[mask, "diameter"] = (
    1329 * 10 ** (-df.loc[mask, "absolute_magnitude"] / 5)
    / np.sqrt(df.loc[mask, "albedo"])
)
```

#### in the same way, filling albedo

```python
mask = (
    df["albedo"].isna()
    & df["absolute_magnitude"].notna()
    & df["diameter"].notna()
    & (df["diameter"] > 0)
)

df.loc[mask, "albedo"] = (
    1329 * 10 ** (-df.loc[mask, "absolute_magnitude"] / 5)
    / df.loc[mask, "diameter"]
) ** 2
```

#### In rows, where the corresponding necessary metric is null, we are not
#### able to determine the missing values using formula, in such case
#### we are filling our null values with medians

```python
df["diameter"] = df["diameter"].fillna(df["diameter"].median())
df["albedo"] = df["albedo"].fillna(df["albedo"].median())
df["rotation_period"] = df["rotation_period"].fillna(df["rotation_period"].median())
```

### Engineered Features


```python
df["orbital_period_years"] = df["semimajor_axis"] ** 1.5
```

#### orbital period years says how long the aesteroids take to orbit the Sun

```python
df["aphelion_distance"] = (
    df["semimajor_axis"] * (1 + df["eccentricity"])
)
```

#### aphelion distance gives us the distance of the asteroid from the sun
#### it helps in comparing the perihelion distance

```python
bins = [-np.inf, 0.1, 0.4, np.inf]
labels = ["circular", "moderate_eccentricity", "high_eccentricity"]
df["size_category"] = pd.cut(df["diameter"], bins=bins, labels=labels)
```

#### Binning is the process of converting the continous variables into discrete intervals
#### by binning the data into circular, moderate eccentricity, and high eccentricity helps in visualization

## Grouping and Aggregation

#### grouping and aggregation helps to summarize the information relatively easily
#### they form the basis for creating tables later on

```python
class_summary = (
    df.groupby("class", dropna=False)
      .agg(
          asteroid_count=("full_name", "count"),
          average_diameter=("diameter", "mean"),
          median_diameter=("diameter", "median"),
          average_albedo=("albedo", "mean")
      )
      .reset_index()
)

print(class_summary)
```

#### this is class-wise information on data of aesteroids
#### this carries essential information, such as MBA class has higher number
#### Aesteroid of class TJN are significantly larger; larger doesn't always means more hazardous

```python
risk_summary = (
    df.groupby(["neo_flag", "pha_flag"], dropna=False)
      .agg(
          asteroid_count=("full_name", "count"),
          median_diameter=("diameter", "median")
      )
      .reset_index()
)
print(risk_summary)
```

##### the risk summary tells us in the data no aesteroids is potentially hazardous
##### the aesteroids which are near Earth has significantly smaller diameter
