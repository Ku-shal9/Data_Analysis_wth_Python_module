## Question 1: Read the CSV file

```python
import pandas as pd

df = pd.read_csv("/content/pandas_dataset.csv")
shape_before = df.shape

df.head()
```

## Question 2: Renaming Columns

```python
## converting to lowercase
df.columns = df.columns.str.lower()

## removing spaces and adding underscore
df.columns = df.columns.str.replace(' ', '_')
print("After: ", df.columns)
```

## Question 3: Drop Unnecessary Rows

```python
df.head()
```

```python
# dropping first empty row
df = df.drop([0])

# reset the index
df.reset_index(drop=True, inplace=True)

df.head()
```

## Question 4: Dropping unnecessary columns

```python
df.drop(columns="customer_name", inplace=True)

"""
customer_name is kind of a unique element, meaning for every customer, there is a unique name that
doesn't contribute much to profit/loss analysis. If all values are unique, we cannot find any patterns whatsoever.

"""

df.head()
```

## Question 5: Missing Values

```python
print(df.isnull().sum())
```

## Question 6: Handle Missing Values

### a. Drop rows based on column

```python
df = df.dropna(subset=['sales']).reset_index(drop=True) # droping rows will change indices thus it is necessary
print(df.isnull().sum())
```

### b. Imputing

```python
df["profit"]=df["profit"].fillna(df["profit"].mean())
df["discount"]=df["discount"].fillna(df["discount"].median())
print(df.isnull().sum())
```

## Question 7: Handling missing categorical data

```python
df["customer_name"] = df["customer_name"].fillna("Unknown")
print(df.isnull().sum())
```

### Question 8: Duplicates

```python
duplicate_values = df.duplicated(subset=["order_id"])

print("Number of duplicated values:", duplicate_values.sum())

before_removal = len(df)

df = df.drop_duplicates(subset=["order_id"])

duplicate_count = df.duplicated(subset=["order_id"]).sum()

after_removing = len(df)

df.reset_index(drop=True, inplace=True)

print("Number of duplicate values removed: ", before_removal-after_removing)

print("Number of duplicates remaining:", duplicate_count)
```

### Question 9: Filtering

```python
# unit_price is greater than 20,000.
df[df["unit_price"] > 20000]
```

```python
# display customer_name, category, and status for orders where unit_price >
# 10,000 and status == "Completed".

df[(df["unit_price"] > 10000) & (df["status"] == "Completed")][["customer_name", "category", "status"]]
```

```python
# Create a new column called total_amount:
df["total_amount"] = df["quantity"] * df["unit_price"] - df["discount"]
df.head()
```

```python
# Create a column called customer_type based on quantity:
import numpy as np

df["customer_type"] = np.where(
    df["quantity"] >= 3,
    "Bulk Buyer",
    "Regular Buyer"
)
```
